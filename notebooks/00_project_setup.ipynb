```python
from pathlib import Path

project_root = Path.cwd() / "document-table-agent"

print("Project root:", project_root)
print("Exists:", project_root.exists())
```

```python
pdf_folder = project_root / "sample_documents"

pdf_files = list(pdf_folder.glob("*.pdf"))

print("PDF files found:")

for pdf in pdf_files:
    print("-", pdf.name)
```

```python
pdf_path = pdf_files[0]

print("Using PDF:")
print(pdf_path)
```

```python
import fitz

doc = fitz.open(pdf_path)

print("PDF opened successfully")
print("Number of pages:", len(doc))
```

```python
for page_number in range(len(doc)):
    page = doc[page_number]
    text = page.get_text()

    print("=" * 70)
    print(f"PAGE {page_number + 1}")
    print("=" * 70)
    print(text[:1000])
    print()
```

```python
search_terms = [
    "Maximum Demand Met",
    "Energy Consumption"
]

for term in search_terms:
    print(f"\nSearching for: {term}")

    for page_number in range(len(doc)):
        text = doc[page_number].get_text()

        if term.lower() in text.lower():
            print(f"Found on page {page_number + 1}")
```

```python
table_searches = {
    "maximum_demand": "Maximum Demand Met",
    "energy_consumption": "Energy Consumption"
}

found_pages = {}

for table_name, search_term in table_searches.items():
    matches = []

    for page_index in range(len(doc)):
        text = doc[page_index].get_text()

        if search_term.lower() in text.lower():
            matches.append(page_index + 1)

    found_pages[table_name] = matches

print(found_pages)
```

```python
energy_page_number = found_pages["energy_consumption"][0]

energy_page = doc[energy_page_number - 1]

energy_text = energy_page.get_text()

print(energy_text)
```

```python
import pdfplumber
```

```python
with pdfplumber.open(pdf_path) as pdf:
    page = pdf.pages[energy_page_number - 1]

    tables = page.extract_tables()

print("Number of tables detected:", len(tables))
```

```python
for table_index, table in enumerate(tables):

    print("\n" + "=" * 100)
    print(f"TABLE {table_index + 1}")
    print("=" * 100)

    print("Number of rows:", len(table))

    if len(table) > 0:
        print("Number of columns:", len(table[0]))

    print("\nFirst 10 rows:\n")

    for row in table[:10]:
        print(row)
```

```python
import pandas as pd
```

```python
for table_index, table in enumerate(tables):

    print(f"\nTABLE {table_index + 1}")

    temp_df = pd.DataFrame(table)

    display(temp_df.head(15))
```

```python
raw_energy_table = tables[0]

raw_energy_df = pd.DataFrame(raw_energy_table)

display(raw_energy_df)
```
