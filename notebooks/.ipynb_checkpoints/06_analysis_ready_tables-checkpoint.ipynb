# Analysis-Ready Long Tables

This notebook runs the profile-driven pipeline and inspects optional electricity-specific long-format outputs. The generic runner remains independent of these postprocessors.

## 1. Project setup and imports

```python
from pathlib import Path
import importlib
import sys

import pandas as pd
from IPython.display import display

working_directory = Path.cwd()
project_root = (
    working_directory
    if (working_directory / "app").exists()
    else working_directory.parent
)
if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))

import app.pipeline.builtin_profiles as builtin_profile_module
import app.pipeline.runner as pipeline_runner
import app.transformation.electricity_analysis as analysis_module

builtin_profile_module = importlib.reload(builtin_profile_module)
pipeline_runner = importlib.reload(pipeline_runner)
analysis_module = importlib.reload(analysis_module)

profiles = builtin_profile_module.BUILTIN_DOCUMENT_PROFILES
run_profiled_pipeline = pipeline_runner.run_profiled_pipeline
reconcile_energy_consumption_totals = (
    analysis_module.reconcile_energy_consumption_totals
)

sample_documents = project_root / "sample_documents"
outputs = project_root / "outputs"
print("Project root:", project_root)
```

## 2. Run the configured PDF pipeline

```python
pdf_files = sorted(sample_documents.glob("*.pdf"))
if not pdf_files:
    raise FileNotFoundError("No PDF files found in sample_documents.")

pipeline_result = run_profiled_pipeline(
    pdf_files[0],
    profiles,
    output_dir=outputs,
    overwrite=True,
)
print("Completed profile:", pipeline_result.profile_name)
```

## 3. Collect profile-derived long tables

```python
energy_result = pipeline_result.tables["energy_consumption"]
maximum_result = pipeline_result.tables["maximum_demand"]
energy_long = energy_result.postprocessed_tables["long"]
maximum_long = maximum_result.postprocessed_tables["long"]

print("Energy long shape:", energy_long.shape)
print("Maximum Demand long shape:", maximum_long.shape)
print("Energy output:", energy_result.postprocessed_output_paths["long"])
print("Maximum output:", maximum_result.postprocessed_output_paths["long"])
```

## 4. Verify schemas, missing values, and uniqueness

```python
checks = pd.DataFrame(
    [
        {
            "table": "Energy Consumption",
            "rows": len(energy_long),
            "columns": energy_long.shape[1],
            "missing cells": int(energy_long.isna().sum().sum()),
            "duplicate keys": int(
                energy_long.duplicated(
                    subset=["Region", "State", "Date"]
                ).sum()
            ),
            "date dtype": str(energy_long["Date"].dtype),
        },
        {
            "table": "Maximum Demand",
            "rows": len(maximum_long),
            "columns": maximum_long.shape[1],
            "missing cells": int(maximum_long.isna().sum().sum()),
            "duplicate keys": int(
                maximum_long.duplicated(
                    subset=["Region", "State", "Date"]
                ).sum()
            ),
            "date dtype": str(maximum_long["Date"].dtype),
        },
    ]
)
display(checks)

assert energy_long.shape == (280, 4)
assert maximum_long.shape == (273, 5)
assert checks["missing cells"].eq(0).all()
assert checks["duplicate keys"].eq(0).all()
```

## 5. Preview analysis-ready data

```python
display(energy_long.head(10))
display(maximum_long.head(10))
display(energy_long.dtypes.to_frame("dtype"))
display(maximum_long.dtypes.to_frame("dtype"))
```

## 6. Reconcile Energy totals

The reported `ALL INDIA` values are compared with the sum of component rows. Small differences are expected from source rounding, so this profile uses a 1.0 MU tolerance.

```python
energy_reconciliation = reconcile_energy_consumption_totals(
    energy_long, tolerance=1.0
)
display(energy_reconciliation)
assert energy_reconciliation["Within_Tolerance"].all()
```

## 7. Findings

- Energy is exported as `Region | State | Date | Energy_Consumption_MU`.
- Maximum Demand is exported as `Region | State | Date | Maximum_Demand_MW | Peak_Shortage_MW`.
- Dates are parsed as datetimes and measures remain numeric.
- Both profile schemas enforce exact row counts, non-null fields, and unique entity/date keys.
- Energy component sums agree with reported totals within the documented rounding tolerance.
- Wide clean CSVs remain available alongside the derived long CSVs.
