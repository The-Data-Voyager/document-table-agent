# Deterministic Table Query Agent

This notebook runs the configured PDF pipeline, registers its analysis-ready tables, and answers structured questions with supporting source rows. Query execution is generic and does not contain electricity-specific calculation logic.

## 1. Project setup and imports

```python
from pathlib import Path
import importlib
import json
import sys

import pandas as pd
from IPython.display import display

working_directory = Path.cwd()
project_root = (
    working_directory
    if (working_directory / "app").exists()
    else working_directory.parent
)
if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))

import app.agent.request_parser as request_parser_module
import app.agent.query_engine as query_engine_module
import app.pipeline.builtin_profiles as builtin_profile_module
import app.pipeline.runner as pipeline_runner

request_parser_module = importlib.reload(request_parser_module)
query_engine_module = importlib.reload(query_engine_module)
builtin_profile_module = importlib.reload(builtin_profile_module)
pipeline_runner = importlib.reload(pipeline_runner)

TableQueryAgent = query_engine_module.TableQueryAgent
profiles = builtin_profile_module.BUILTIN_DOCUMENT_PROFILES
run_profiled_pipeline = pipeline_runner.run_profiled_pipeline

sample_documents = project_root / "sample_documents"
print("Project root:", project_root)
```

## 2. Run the profile-driven pipeline

```python
pdf_files = sorted(sample_documents.glob("*.pdf"))
if not pdf_files:
    raise FileNotFoundError("No PDF files found in sample_documents.")

pipeline_result = run_profiled_pipeline(pdf_files[0], profiles)
analysis_tables = {
    table_name: table_result.postprocessed_tables["long"]
    for table_name, table_result in pipeline_result.tables.items()
}
agent = TableQueryAgent(analysis_tables)
print("Completed profile:", pipeline_result.profile_name)
print("Registered tables:", sorted(analysis_tables))
```

## 3. Inspect the available query schemas

```python
schemas = agent.describe_tables()
schema_summary = pd.DataFrame(
    [
        {"Table": name, "Rows": details["rows"], "Columns": details["columns"]}
        for name, details in schemas.items()
    ]
)
display(schema_summary)
assert schemas["energy_consumption"]["rows"] == 280
assert schemas["maximum_demand"]["rows"] == 273
```

## 4. Top five states by weekly energy consumption

The reported ALL INDIA total is excluded before state-level grouping.

```python
top_energy_request = {
    "table": "energy_consumption",
    "filters": [
        {"column": "Region", "operator": "ne", "value": "ALL INDIA"}
    ],
    "group_by": ["Region", "State"],
    "aggregations": [
        {
            "column": "Energy_Consumption_MU",
            "operation": "sum",
            "alias": "Weekly_Energy_MU",
        }
    ],
    "order_by": [{"column": "Weekly_Energy_MU", "direction": "desc"}],
    "limit": 5,
}
top_energy = agent.ask(top_energy_request)
display(top_energy.answer)
print("Supporting source rows:", top_energy.matched_row_count)
assert top_energy.returned_row_count == 5
assert top_energy.matched_row_count == 273
```

### Evidence preview

`evidence` contains every filtered source row used to calculate and rank the grouped answer.

```python
display(top_energy.evidence.head(10))
```

## 5. Daily maximum demand for Delhi and Haryana

```python
state_demand = agent.ask(
    {
        "table": "maximum_demand",
        "columns": [
            "Region",
            "State",
            "Date",
            "Maximum_Demand_MW",
            "Peak_Shortage_MW",
        ],
        "filters": [
            {"column": "State", "operator": "in", "value": ["Delhi", "Haryana"]},
            {
                "column": "Date",
                "operator": "between",
                "value": ["2026-03-30", "2026-04-05"],
            },
        ],
        "order_by": [
            {"column": "State", "direction": "asc"},
            {"column": "Date", "direction": "asc"},
        ],
        "limit": 20,
    }
)
display(state_demand.answer)
assert state_demand.returned_row_count == 14
```

## 6. Maximum observed peak shortage by region

```python
regional_shortage = agent.ask(
    {
        "table": "maximum_demand",
        "group_by": ["Region"],
        "aggregations": [
            {
                "column": "Peak_Shortage_MW",
                "operation": "max",
                "alias": "Maximum_Peak_Shortage_MW",
            }
        ],
        "order_by": [
            {"column": "Maximum_Peak_Shortage_MW", "direction": "desc"}
        ],
    }
)
display(regional_shortage.answer)
```

## 7. Submit the same contract as JSON

A future natural-language adapter only needs to produce this validated request format.

```python
json_request = json.dumps(
    {
        "table": "energy_consumption",
        "filters": [
            {"column": "Region", "operator": "ne", "value": "ALL INDIA"}
        ],
        "aggregations": [
            {
                "column": "Energy_Consumption_MU",
                "operation": "mean",
                "alias": "Average_State_Daily_Energy_MU",
            }
        ],
    }
)
average_energy = agent.ask(json_request)
display(average_energy.answer)
print(json_request)
```

## 8. Findings

- The agent queries flat analysis tables through a generic schema-driven engine.
- Requests can filter strings, numbers, and parsed dates; project columns; group and aggregate; sort; and limit results.
- Unsupported fields, operators, tables, and columns fail explicitly. No arbitrary Pandas or Python expression is evaluated.
- Every answer includes the filtered source rows used as evidence, while registered DataFrames remain unchanged.
- Natural-language or LLM interpretation can be added later as a separate adapter that emits the same strict request contract.
