# Table Validation

This notebook validates raw extracted tables without cleaning or transforming them. The workflow preserves the project boundary: **Extraction → Validation → Transformation**.

## 1. Project setup and imports

```python
from pathlib import Path
import sys

import pandas as pd
from IPython.display import display

working_directory = Path.cwd()
project_root = (
    working_directory
    if (working_directory / "app").exists()
    else working_directory.parent
)

if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))

sample_documents = project_root / "sample_documents"

from app.extraction.table_extractor import extract_table_as_dataframe
from app.parsers.pdf_parser import search_pdf
from app.validation.table_validator import validate_table

print("Project root:", project_root)
print("Sample documents exists:", sample_documents.exists())
```

## 2. Load the sample PDF

```python
pdf_files = sorted(sample_documents.glob("*.pdf"))

if not pdf_files:
    raise FileNotFoundError("No PDF files found in sample_documents.")

pdf_path = pdf_files[0]
print("Using PDF:", pdf_path)
```

## 3. Locate Energy Consumption with `search_pdf()`

```python
energy_pages = search_pdf(pdf_path, "Energy Consumption")

if not energy_pages:
    raise ValueError("Energy Consumption was not found in the sample PDF.")

energy_page_number = energy_pages[0]
print("Energy Consumption pages:", energy_pages)
```

## 4. Extract the raw Energy Consumption table

```python
raw_energy_df = extract_table_as_dataframe(
    pdf_path=pdf_path,
    page_number=energy_page_number,
    table_index=0,
)

print("Raw Energy table shape:", raw_energy_df.shape)
display(raw_energy_df.head())
```

## 5. Validate and display the Energy Consumption report

```python
energy_before_validation = raw_energy_df.copy(deep=True)
energy_report = validate_table(raw_energy_df)
pd.testing.assert_frame_equal(
    raw_energy_df,
    energy_before_validation,
    check_exact=True,
)

display(pd.Series(energy_report, name="value").to_frame())
```

## 6. Locate and extract Maximum Demand / Peak Shortage

```python
maximum_demand_pages = search_pdf(pdf_path, "Maximum Demand Met")

if not maximum_demand_pages:
    raise ValueError("Maximum Demand Met was not found in the sample PDF.")

maximum_demand_page_number = maximum_demand_pages[0]
raw_maximum_demand_df = extract_table_as_dataframe(
    pdf_path=pdf_path,
    page_number=maximum_demand_page_number,
    table_index=0,
)

print("Maximum Demand pages:", maximum_demand_pages)
print("Raw Maximum Demand table shape:", raw_maximum_demand_df.shape)
display(raw_maximum_demand_df.head())
```

## 7. Validate and display the Maximum Demand report

```python
maximum_before_validation = raw_maximum_demand_df.copy(deep=True)
maximum_demand_report = validate_table(raw_maximum_demand_df)
pd.testing.assert_frame_equal(
    raw_maximum_demand_df,
    maximum_before_validation,
    check_exact=True,
)

display(pd.Series(maximum_demand_report, name="value").to_frame())
```

## 8. Compare validation results

```python
comparison_fields = [
    "rows",
    "columns",
    "missing_cells",
    "duplicate_rows",
    "empty_rows",
    "empty_columns",
    "possible_title_rows",
    "possible_header_rows",
    "date_like_cells",
    "possible_hierarchical_header",
    "possible_merged_cell_columns",
]

comparison = pd.DataFrame(
    {
        "Energy Consumption": {
            field: energy_report[field] for field in comparison_fields
        },
        "Maximum Demand / Peak Shortage": {
            field: maximum_demand_report[field]
            for field in comparison_fields
        },
    }
)

display(comparison)
```

## 9. Findings

- Validation is observational: both raw DataFrames are asserted identical before and after validation.
- Missing and blank cells are reported, not filled. Internal missing runs can suggest vertically merged group labels, but ordinary missing data can produce the same pattern.
- Title and header rows are heuristic candidates. The Maximum Demand table is expected to produce adjacent header candidates, which is reported cautiously as a possible hierarchical header.
- Date matching accepts only complete, valid date patterns; date-like values can occur in either headers or data.
- No state-name cleaning, bilingual-text removal, header flattening, reshaping, numeric conversion, or other transformation is performed.
